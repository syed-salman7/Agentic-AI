```python
import os
from getpass import getpass

api_key = getpass("Enter your Gemini API key: ")

os.environ["GOOGLE_API_KEY"] = api_key

print("API key configured successfully!")
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings

embeddings = GoogleGenerativeAIEmbeddings(
    model="models/gemini-embedding-001"
)

print("Embedding model created successfully!")
```

```python
from langchain_core.documents import Document

documents = [
    Document(
        page_content="""
        Machine Learning is a subset of Artificial Intelligence.
        It allows computers to learn patterns from data and make
        predictions or decisions without being explicitly programmed
        for every task.
        """
    ),

    Document(
        page_content="""
        Deep Learning is a subset of Machine Learning that uses
        neural networks with multiple layers. It is commonly used
        for image recognition, speech recognition, and natural
        language processing.
        """
    ),

    Document(
        page_content="""
        Generative AI refers to artificial intelligence systems
        that can create new content such as text, images, audio,
        video, and computer code.
        """
    ),

    Document(
        page_content="""
        Large Language Models are AI models trained on large
        amounts of text data. They can understand and generate
        human-like text.
        """
    ),

    Document(
        page_content="""
        Retrieval Augmented Generation combines information
        retrieval with language generation. A RAG system retrieves
        relevant information from a knowledge source and provides
        it to a language model to generate a grounded answer.
        """
    )
]

print("Documents created:", len(documents))
```

```python
from langchain_community.vectorstores import Chroma

vectorstore = Chroma.from_documents(
    documents=documents,
    embedding=embeddings,
    collection_name="semantic_search_demo"
)

print("Vector database created successfully!")
```

```python
def semantic_search(query, k=3):
    results = vectorstore.similarity_search(
        query,
        k=k
    )

    print("Query:", query)
    print("\nMost relevant documents:\n")

    for i, doc in enumerate(results, 1):
        print(f"--- Result {i} ---")
        print(doc.page_content.strip())
        print()
```

```python
semantic_search(
    "How do computers learn from data?"
)
```

```python
semantic_search(
    "How can AI create new content?"
)
```

```python
semantic_search(
    "How do computers learn patterns from examples?"
)
```

```python
query = "How can AI create new content?"

results = vectorstore.similarity_search_with_score(
    query,
    k=5
)

print("Query:", query)
print("\nResults:\n")

for i, (doc, score) in enumerate(results, 1):
    print(f"--- Result {i} ---")
    print("Score:", score)
    print("Document:")
    print(doc.page_content.strip())
    print()
```
